# Logical Operators

```python
## And Operator


X=True

Y=True

result= X and Y
print(result)  # Output: True
```

```python
X=True

Y=False

result= X and Y
print(result) # Output: False
```

```python
# Or Operator

X=True

Y=True

result= X or Y
print(result) 
```

```python
X=True

Y=False

result= X or Y
print(result)
```
